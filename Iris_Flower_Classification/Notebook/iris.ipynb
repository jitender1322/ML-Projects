```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
```

```python
iris = load_iris()

df = pd.DataFrame(iris.data,columns=iris.feature_names)
df
```

```python
df['species'] = iris.target
```

```python
df.sample()
```

```python
df['species']=df['species'].map({
    0: "setosa",
    1: "versicolor",
    2: "virginica"
})
```

```python
x = df.drop("species",axis=1)
y= df["species"]
```

```python
x.sample()
y.sample()
```

```python
from sklearn.model_selection import train_test_split


X_train,X_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

### Logistic Regression

Logistic Regression is a supervised machine learning algorithm used for **classification problems**.

Even though the name contains "regression", it is mainly used for **predicting categorical outcomes**.

In this project, Logistic Regression will predict the **species of an iris flower** based on:

- sepal length
- sepal width
- petal length
- petal width

Logistic Regression works by estimating the **probability that a data point belongs to a class**.

For binary classification it uses the **sigmoid function** to map predictions between 0 and 1.

For multi-class problems like the Iris dataset, it uses **softmax** internally.

Advantages:
- Simple and fast
- Works well with linearly separable data
- Good baseline model

```python
from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression()
log_reg.fit(X_train, y_train)

y_pred_lr = log_reg.predict(X_test)
y_pred_lr
```

```python
from sklearn.metrics import accuracy_score

accuracy_lr = accuracy_score(y_test, y_pred_lr)
```

### Classification Report

Accuracy only tells us how many predictions were correct overall.

However, in classification problems we often need more detailed evaluation metrics.

The classification report provides:

• Precision  
• Recall  
• F1-score  
• Support  

These metrics help us understand how well the model performs for each class.

```python
from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred_lr)
print(report)
```

### K-Nearest Neighbors (KNN)

K-Nearest Neighbors (KNN) is a supervised machine learning algorithm used for both classification and regression problems.

In classification problems, KNN predicts the class of a data point based on the majority class among its nearest neighbors.

The algorithm works by calculating the distance between the new data point and all training data points. Then it selects the K closest points and assigns the class that appears most frequently among them.

Key characteristics of KNN:
- It is a distance-based algorithm.
- It does not build a model during training (lazy learner).
- It works well with small datasets.

In this project, KNN will classify iris flowers based on similarities in their petal and sepal measurements.

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=3)

knn.fit(X_train, y_train)

y_pred_knn = knn.predict(X_test)
```

```python
accuracy_knn = accuracy_score(y_test, y_pred_knn)

accuracy_knn
print(classification_report(y_test, y_pred_knn))
```

### Decision Tree Classifier

Decision Tree is a supervised machine learning algorithm used for both classification and regression problems.

It works by splitting the dataset into smaller subsets based on feature values.  
The algorithm creates a tree-like structure where each node represents a decision rule based on a feature.

For example, the model may create rules such as:

if petal_length < 2.5 → Setosa  
else → check another feature

The tree keeps splitting the data until it reaches a final prediction at the leaf nodes.

Advantages of Decision Trees:
- Easy to understand and interpret
- Works with both numerical and categorical data
- Does not require feature scaling

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier()
dt.fit(X_train, y_train)
y_pred_dt = dt.predict(X_test)
```

```python
accuracy_dt = accuracy_score(y_test, y_pred_dt)

accuracy_dt
print(classification_report(y_test, y_pred_dt))
```

### Why Random Forest Works Better Than a Single Decision Tree

A single Decision Tree can easily overfit the training data because it may learn very specific patterns.

Random Forest solves this problem by creating many decision trees using different subsets of the data.

Each tree makes its own prediction, and the final prediction is determined by majority voting.

Because many trees participate in the decision, the model becomes more stable and less sensitive to noise in the dataset.

This technique is called **Ensemble Learning**, where multiple weak models combine to create a stronger model.

### Random Forest Classifier

Random Forest is an ensemble learning algorithm that combines multiple decision trees to improve prediction accuracy and reduce overfitting.

Each tree is trained on a random subset of the dataset, and the final prediction is determined by majority voting among all trees.

Random Forest usually performs better than a single Decision Tree because it reduces variance and improves generalization.

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier()

rf.fit(X_train,y_train)

y_pred_rf = rf.predict(X_test)
```

```python
accuracy_rf = accuracy_score(y_test, y_pred_rf)

accuracy_rf
print(classification_report(y_test, y_pred_rf))
```

### Support Vector Machine (SVM)

Support Vector Machine is a supervised machine learning algorithm used for classification and regression problems.

SVM works by finding the optimal decision boundary that separates different classes in the dataset.

The algorithm tries to maximize the margin between the classes. The margin is the distance between the decision boundary and the closest data points from each class.

These closest data points are called support vectors.

SVM is particularly effective in high-dimensional datasets and can handle both linear and non-linear classification problems.

```python
from sklearn.svm import SVC

svm = SVC()
svm.fit(X_train,y_train)
y_pred_svm = svm.predict(X_test)
```

```python
accuracy_svm = accuracy_score(y_test, y_pred_svm)

accuracy_svm

print(classification_report(y_test, y_pred_svm))
```

### Confusion Matrix

A confusion matrix is used to evaluate the performance of a classification model.

It shows the number of correct and incorrect predictions for each class.

Rows represent the actual classes and columns represent the predicted classes.

The diagonal values represent correct predictions, while the off-diagonal values represent misclassifications.

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns

cm = confusion_matrix(y_test, y_pred_rf)

cm
```

```python
sns.heatmap(cm, annot=True, cmap="Blues")

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```
