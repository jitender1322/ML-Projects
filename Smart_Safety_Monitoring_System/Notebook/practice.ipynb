```python
cap = cv2.VideoCapture(0)
start = time.perf_counter()
frames = 0
while True:
    ret,frame = cap.read()
    if ret:
        cv2.imshow("webcame",frame)
    key = cv2.waitKey(1)
    frames+=1

    if key==ord("q"):
        break
end = time.perf_counter()

total = end - start

print(frames/total)

cap.release()
cv2.destroyAllWindows()
```

```python
cap = cv2.VideoCapture(0)

prev_time = time.perf_counter()

while True:
    ret, frame = cap.read()

    if ret:
        current_time = time.perf_counter()

        fps = 1 / (current_time - prev_time)

        prev_time = current_time

        cv2.putText(
            frame,
            f"FPS: {fps:.2f}",
            (20, 40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0, 255, 0),
            2
        )

        cv2.imshow("Webcam", frame)

    key = cv2.waitKey(1)

    if key == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
cap = cv2.VideoCapture(0)

prev_time = time.perf_counter()

alpha = 0.5
smoothed_fps = 0

while True:
    ret,frame = cap.read()

    if ret:
        current_time = time.perf_counter()
        current_fps = 1/(current_time-prev_time)
        prev_time=current_time

        # EMA - Exponential Moving Average

        if smoothed_fps==0:
            smoothed_fps=current_fps
        else:
            smoothed_fps = alpha+current_fps+(1-alpha)*smoothed_fps

        cv2.putText(frame,f"FPS: {smoothed_fps:.2f}",(20, 40),cv2.FONT_HERSHEY_SIMPLEX, 1,(0, 255, 0),2)

        cv2.imshow("web",frame)

    key = cv2.waitKey(1)
    if key == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
cap = cv2.VideoCapture(0)

start_time = time.perf_counter()
frame_count = 0
fps = 0

while True:

    ret, frame = cap.read()

    if ret:
        frame_count += 1

        current_time = time.perf_counter()
        elapsed = current_time - start_time

        if elapsed >= 1:
            fps = frame_count / elapsed

            frame_count = 0
            start_time = current_time

        cv2.putText(
            frame,
            f"FPS: {fps:.2f}",
            (20, 40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0, 255, 0),
            2
        )

        cv2.imshow("Webcam", frame)

    key = cv2.waitKey(1)

    if key == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
from ultralytics import YOLO

model = YOLO("yolo26n.pt")
model
```

```python
results = model("../Images/img.jpg")
print(type(results))
print(len(results))
print(type(results[0]))
print(results[0].boxes)
```

```python
image = cv2.imread("../Images/img.jpg")
for i in range(len(result.boxes)):
    box = result.boxes.xyxy[i]
    conf = result.boxes.conf[i]
    cls = result.boxes.cls[i]

    class_name = result.names[int(cls)]
    conf = result.boxes.conf[i]
    label = f"{class_name} {conf:.2f}"

    box = box.cpu().numpy()

    x1, y1, x2, y2 = map(int, box)

    cv2.rectangle(
        image,
        (x1, y1),
        (x2, y2),
        (0, 255, 0),
        2
    )

    cv2.putText(
        image,
        label,
        (x1, y1 - 10),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.6,
        (0, 255, 0),
        2
    )

    cv2.imshow(f"Detection+{i}", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import time

cap = cv2.VideoCapture(0)

while True:
    ret,frame = cap.read()

    if not ret:
        break

    start = time.perf_counter()
    results = model(frame,conf=0.5,imgsz=320)
    end = time.perf_counter()

    inference_time = end-start
    print(inference_time)

    fps = 1/inference_time
    print(fps)

    result = results[0]
    annotated_frame = result.plot()

    cv2.imshow("webcam",frame)
    cv2.imshow("web",annotated_frame)

    key = cv2.waitKey(1)

    if key==ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
