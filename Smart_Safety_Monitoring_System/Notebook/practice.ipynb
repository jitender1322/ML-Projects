```python
cap = cv2.VideoCapture(0)
start = time.perf_counter()
frames = 0
while True:
    ret,frame = cap.read()
    if ret:
        cv2.imshow("webcame",frame)
    key = cv2.waitKey(1)
    frames+=1

    if key==ord("q"):
        break
end = time.perf_counter()

total = end - start

print(frames/total)

cap.release()
cv2.destroyAllWindows()
```

```python
cap = cv2.VideoCapture(0)

prev_time = time.perf_counter()

while True:
    ret, frame = cap.read()

    if ret:
        current_time = time.perf_counter()

        fps = 1 / (current_time - prev_time)

        prev_time = current_time

        cv2.putText(
            frame,
            f"FPS: {fps:.2f}",
            (20, 40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0, 255, 0),
            2
        )

        cv2.imshow("Webcam", frame)

    key = cv2.waitKey(1)

    if key == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
cap = cv2.VideoCapture(0)

prev_time = time.perf_counter()

alpha = 0.5
smoothed_fps = 0

while True:
    ret,frame = cap.read()

    if ret:
        current_time = time.perf_counter()
        current_fps = 1/(current_time-prev_time)
        prev_time=current_time

        # EMA - Exponential Moving Average

        if smoothed_fps==0:
            smoothed_fps=current_fps
        else:
            smoothed_fps = alpha+current_fps+(1-alpha)*smoothed_fps

        cv2.putText(frame,f"FPS: {smoothed_fps:.2f}",(20, 40),cv2.FONT_HERSHEY_SIMPLEX, 1,(0, 255, 0),2)

        cv2.imshow("web",frame)

    key = cv2.waitKey(1)
    if key == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
cap = cv2.VideoCapture(0)

start_time = time.perf_counter()
frame_count = 0
fps = 0

while True:

    ret, frame = cap.read()

    if ret:
        frame_count += 1

        current_time = time.perf_counter()
        elapsed = current_time - start_time

        if elapsed >= 1:
            fps = frame_count / elapsed

            frame_count = 0
            start_time = current_time

        cv2.putText(
            frame,
            f"FPS: {fps:.2f}",
            (20, 40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0, 255, 0),
            2
        )

        cv2.imshow("Webcam", frame)

    key = cv2.waitKey(1)

    if key == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

### ROI = Region of Interest

```python
cap = cv2.VideoCapture(0)

while True:
    ret,frame = cap.read()

    if ret:
        roi = frame[100:380,150:500]
        cv2.imshow("full",frame)
        cv2.imshow("roi",roi)

        key = cv2.waitKey(1)

        if key==ord("q"):
            break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()

    if ret:
        roi = frame[100:380, 150:500]

        cv2.rectangle(
            frame,
            (150, 100),
            (500, 380),
            (0, 255, 0),
            2
        )

        cv2.imshow("Webcam", frame)
        cv2.imshow("ROI", roi)

    key = cv2.waitKey(1)

    if key == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
